- 1.Build silver drivers dimension
- 2.Build gold ref_nationality_region table
- 3.Join the data from constructors with ref_nationality_region using nationality
- 4.select the required columns
- 5.Write transformed data to gold dim_drivers table

```python
target_table=f"{catalog_name}.{gold_schema}.dim_drivers"
```

```python
import pyspark.sql.functions as F
```

## Step -1 Read source tables
- silver.drivers
- gold.ref_nationality_region

```python
drivers_df=(
    spark.table(f"{catalog_name}.{silver_schema}.drivers").filter(F.col("batch_id") == v_batch_id)
)
```

```python

ref_nationality_region_df=spark.table(f"{catalog_name}.{gold_schema}.ref_nationality_region")
```

**Step 2 Join the contructors with ref_nationality_region using nationality**

```python
dim_drivers_df=(
    drivers_df
    .join(
        ref_nationality_region_df,
        drivers_df.nationality==ref_nationality_region_df.nationality,
        "left"
    )
    .select(
        drivers_df.driver_id,
        drivers_df.driver_name,
        drivers_df.date_of_birth,
        drivers_df.nationality,
        ref_nationality_region_df.region.alias('nationality_region')
        
    )
)
```

```python
write_to_gold(
    dim_drivers_df,
    target_table,
    merge_condition="s.driver_id=t.driver_id",
    columns_to_update=[
        "driver_name",
        "date_of_birth",
        "nationality",
        "nationality_region"  
    ]
)
```
