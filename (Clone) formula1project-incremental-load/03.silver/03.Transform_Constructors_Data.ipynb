## Transform Constructors Data
- Read bronze Constructors table
- Keep only the columns required for analytics(Drop url column)
- Standardise column name using snakecase(constructorId->constructor_id)
- Rename columns to make them more meaningful(name->constructor_name)
- Remove duplicate records.
- Transform values of column nationality to Title Case
- Write the transformed data to silver constructor table

```python
dbutils.widgets.text("p_batch_id","")
v_batch_id=dbutils.widgets.get("p_batch_id")
```

```python
bronze_table=f"{catalog_name}.{bronze_schema}.constructors"
silver_table=f"{catalog_name}.{silver_schema}.constructors"
```

```python
import pyspark.sql.functions as F
```

### Step 1 Read bronze circuits table

```python
constructors_df=(
    spark.table(bronze_table).filter(F.col("batch_id") == v_batch_id)
)
```

### Srep 2 -Keep only the columns required for analytics(Drop url column)

```python
#flexible   `

constructors_selected_df=(
    constructors_df.drop("url")
    
)
```

### Step 3 & 4 - Stamdardise column name
- Standardise column name using snakecase(constructorId->constructor_id)
- Rename columns to make them more meaningful(name->constructor_name)

```python
constructors_renamed_df=(
    constructors_selected_df.withColumnRenamed("constructorId","constructor_id")
                       .withColumnRenamed("name","constructor_name")
                )
display(constructors_renamed_df)
```

### Step -5 Remove duplicates

```python
#dropduplicates - you can give column name and distinct will be applied on that particular column only
constructors_distinct_df=constructors_renamed_df.dropDuplicates(["constructor_id"])
display(constructors_distinct_df)
```

### Step -6 Transform values of column nationality to Title Case

```python
constructor_final_df=(
    constructors_distinct_df
    .withColumn("nationality",F.initcap(F.col("nationality")))
)
#
```

```python
display(constructor_final_df)
```

### Step-8 Write the transformed data to silver circuits table

```python
write_to_silver(
    constructor_final_df,
    silver_table,
    merge_condition="s.constructor_id=t.constructor_id",
    columns_to_update=[
        "constructor_id",
        "constructor_name",
        "nationality",
        "ingestion_timestamp",
        "source_file",
        "batch_id"
    ]
)
```
