# Creating external location

```python
%fs ls 'abfss://formula1@databricksstorageacc524.dfs.core.windows.net/landing'
```

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricksstorageacc524_formula1
URL 'abfss://formula1@databricksstorageacc524.dfs.core.windows.net/' 
WITH (STORAGE CREDENTIAL databricks_project)
COMMENT 'External location pointing to Azure ADLS Gen2';
```

```python
%sql
CREATE CATALOG IF NOT EXISTS formula1
  MANAGED LOCATION 'abfss://formula1@databricksstorageacc524.dfs.core.windows.net/' 
  COMMENT 'This is catalog for formula1 project';

  SHOW CATALOGS ;

```

# Creating Schemas
