## Build Driver Standings
**Sources**
- 1.fact_session_results
- 2.dim_constructors
- **Output Columns**
- 1.season
- 2.constructor id
- 3.constructor name
- 4.nationality
- 5.race starts
- 6.total points
- 7.number of wins
- 8.number of podiums
- 9.standing position

```python
%sql
CREATE OR REPLACE VIEW formula1.gold.vw_constructors_standings
AS
with constructor_session_summary AS (
Select r.season,
       d.constructor_id,
       d.constructor_name,
       d.nationality,
       count(*) AS race_starts,
       sum(r.points) AS total_points,
       count_if(r.is_win) AS number_of_wins,
       count_if(r.is_podium) AS number_of_podiums
from formula1_incr.gold.fact_session_results r
join formula1_incr.gold.dim_constructors d
on r.constructor_id=d.constructor_id
group by r.season,
       d.constructor_id,
       d.constructor_name,
       d.nationality
)
Select season,
       constructor_id,
       constructor_name,
       nationality,
        rank() OVER(PARTITION BY season order by total_points desc,number_of_wins desc ) AS standings,
       race_starts,
       total_points,
       number_of_wins,
       number_of_podiums
      
       
from constructor_session_summary
```

```python
%sql
select * from formula1_incr.gold.vw_constructors_standings 
where season =2025
```
