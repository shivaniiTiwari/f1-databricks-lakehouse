# Creating external location

```python
%fs ls 'abfss://formula1-incr@databricksstorageacc524.dfs.core.windows.net/landing'
```

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricksstorageacc524_formula1_incr
URL 'abfss://formula1-incr@databricksstorageacc524.dfs.core.windows.net/' 
WITH (STORAGE CREDENTIAL databricks_project)
COMMENT 'External location for formula1-incr container';
```

```python
%sql
CREATE CATALOG IF NOT EXISTS formula1_incr
  MANAGED LOCATION 'abfss://formula1-incr@databricksstorageacc524.dfs.core.windows.net/' 
  COMMENT 'This is catalog for formula1 incr project';

  SHOW CATALOGS ;

```
