## Transform Races Data
- Read bronze Races table
- Keep only the columns required for analytics(Drop url column)
- Standardise column name using snakecase(racesId->races_id)
- Rename columns to make them more meaningful(dates->race_date)
- Remove duplicate records.
- Transform values of column race_name  to Title Case
- Write the transformed data to silver races table

```python
dbutils.widgets.text("p_batch_id","")
v_batch_id=dbutils.widgets.get("p_batch_id")
```

```python
bronze_table=f"{catalog_name}.{bronze_schema}.races"
silver_table=f"{catalog_name}.{silver_schema}.races"
```

```python
%run ../00-common/03.silver-help
```

```python
import pyspark.sql.functions as F
```

### Step 1 Read bronze races table

```python
races_df=(
    spark.table(bronze_table).filter(F.col("batch_id") == v_batch_id)
)
```

### Srep 2 -Keep only the columns required for analytics(Drop url column)

```python
#flexible   `

races_selected_df=races_df.select(
    F.col("season"),
    F.col("round"),
    F.col("raceName"),
    F.col("date"),
    F.col("circuitId"),
    F.col("ingestion_timestamp"),
    F.col("source_file"),
    F.col("batch_id")
)
#display(races_selected_df)

               
```

### Step 3 & 4 - Stamdardise column name
- Standardise column name using snakecase(circuitsId->circuits_id)
- Rename columns to make them more powerful(lat->latitiude)

```python
races_renamed_df=(
    races_selected_df
    .withColumnRenamed("circuitId","circuit_id")
    .withColumnRenamed("raceName","race_Name")
    .withColumnRenamed("date","race_date")
)
#display(races_renamed_df)
```

### Step -5 Remove duplicates

```python
#dropduplicates - you can give column name and distinct will be applied on that particular column only
races_distinct_df=races_renamed_df.dropDuplicates(["season","round"])
#display(races_distinct_df)
```

### Step -6 Transform values of column circuit_name and locality to Title Case

```python
races_final_df=(
    races_distinct_df
    .withColumn("race_name",F.initcap(F.col("race_name")))
)
#display(races_final_df)
```

### Step-7 Write the transformed data to silver circuits table
