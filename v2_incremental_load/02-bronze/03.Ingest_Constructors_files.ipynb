###  Step 1- Read the csv file using dataframe reader API

```python

#schema declaration using ddl
constructors_schema= """constructorId STRING,
  name STRING,
  nationality STRING,
  url STRING"""
```

```python
#spark reader API
constructor_df=(spark.read
                .format('json')
                .schema(constructors_schema)
                .option("mode","failFast")
                .load(source_file)
                )
```
